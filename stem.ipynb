A Stem plot represents data points using vertical lines (stems) and markers.

Unlike a normal line plot, it emphasizes individual discrete observations.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6]
y = [10, 15, 8, 20, 12, 25]

plt.stem(x, y)

plt.xlabel("Time")
plt.ylabel("Values")
plt.title("Stem Plot")

plt.show()
```
