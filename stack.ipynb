A stackplot() is used to show how different categories contribute to a total value over a period of time.

```python

import numpy as np
import matplotlib.pyplot as plt

# Months
months = np.arange(1, 7)

# Sales data
electronics = [20, 25, 30, 28, 35, 40]
clothing = [15, 18, 20, 22, 25, 30]
groceries = [40, 45, 50, 55, 60, 70]

# Create figure
plt.figure(figsize=(11, 6))

# Stacked area chart
plt.stackplot(
    months,
    electronics,
    clothing,
    groceries,
    labels=["Electronics", "Clothing", "Groceries"],
    alpha=0.8
)

# Formatting
plt.title("Monthly Sales Contribution by Category")
plt.xlabel("Month")
plt.ylabel("Sales (₹ Thousands)")
plt.xticks(months)
plt.legend(loc="upper left")
plt.grid(alpha=0.2)

plt.tight_layout()
plt.show()
```

```python
##Important parameters
""" Parameter	Purpose
    x--->	X-axis values
   *y--->	Multiple datasets
   labels--->	Category names
   colors--->	Area colors
   alpha--->	Transparency
   baseline--->	Starting baseline"""
```
