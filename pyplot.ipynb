```python
import matplotlib.pyplot as plt
#plt.pie(value,labels=label name, colors=color name, autopct='%1.1f%%')

'''The string '%1.1f%%' is built out of distinct formatting commands:
% (First): Tells Python to start a string formatting operation.
1 (Before the decimal): Specifies the minimum width of the total string. Because "1" is smaller than the actual length of the numbers (like 25.0), Python simply ignores it and grows the string to whatever size it actually needs.
.1 (After the decimal): Sets the precision. This forces the percentage to round to exactly one decimal place (e.g., 25.34% becomes 25.3%).
f: Specifies that the number should be treated as a floating-point (decimal) number.
%% (End): Escapes the percentage symbol so that a physical % character is printed right after the number'''

region=['north','south','east','west']
revenue=[25000,15000,20000,10000]

plt.pie(revenue,labels=region,colors=['red','blue','green','orange'],autopct='%1.1f%%')

plt.title('Revenue contribution by region')
plt.show()
```

```python
#plt.subplots(nrows, ncols ,index)
import matplotlib.pyplot as plt
x=[1,2,3,4]
y=[10,20,15,25]

plt.subplot(1,2,1)#1row, 2 columns, 1st plot
plt.plot(x,y)
plt.title('Line Plot')

plt.subplot(1,2,2)#1row, 2 columns, 2nd plot
plt.bar(x,y)
plt.title('Bar Plot')

plt.show()
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4]
y=[10,20,15,25]

#fig, ax = plt.subplots(1,2, figsize=(10,5)) #1 row, 2 columns

fig, ax = plt.subplots(1,2, figsize=(10,5)) #1 row, 2 columns
ax[0].plot(x,y) 
ax[0].set_title('Line Plot')

ax[1].bar(x,y)
ax[1].set_title('Bar Plot')

fig.suptitle('Comparison of Line and Bar Plots') # Overall title for the figure

plt.tight_layout() # Adjusts the spacing between subplots to prevent overlap

plt.show()
```
