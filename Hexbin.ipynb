A Hexbin plot displays the density of data points using hexagonal bins.

It's especially useful when you have thousands of data points overlapping in a scatter plot.

```python

import numpy as np
import matplotlib.pyplot as plt

# Generate sample data
np.random.seed(42)

x = np.random.normal(50, 15, 5000)
y = 2 * x + np.random.normal(0, 20, 5000)

# Create Hexbin plot
plt.figure(figsize=(10, 6))

hb = plt.hexbin(
    x,
    y,
    gridsize=30,
    cmap="viridis",
    mincnt=1
)

# Add colorbar
plt.colorbar(hb, label="Number of Data Points")

# Labels and title
plt.xlabel("Study Hours")
plt.ylabel("Performance Score")
plt.title("Hexbin Plot - Study Hours vs Performance")

plt.tight_layout()
plt.show()
```
