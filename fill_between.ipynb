```python
import matplotlib.pyplot as plt
```

```python
#Fills the region between curves or between a curve and a baseline.
#plt.fill_between(x, y1, y2=0, where=None, alpha=None)



x = [1, 2, 3, 4, 5] 
low = [2, 3, 4, 5, 6]
high = [4, 5, 6, 8, 9]
plt.plot(x, low) 
plt.plot(x, high ) 
plt.fill_between(x, low, high, alpha=0.2) 
plt.legend() 
plt.show()
```
