```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import PillowWriter
import matplotlib.animation as FuncAnimation




t= np.linspace(0,10,100)
y= np.sin(t)



fig , ax=plt.subplots()



ax.set_xlim([min(t),max(t)])
ax.set_ylim([-2,2])

# comma after the variable animated_plot is used for tuple unpacking to extract the single line object out of the list returned by Matplotlib's ax.plot()
animated_plot,=ax.plot([],[])
print("animated_plot",animated_plot)

def update_data(frame):
    animated_plot.set_data(t[:frame],y[:frame])

    return animated_plot,

#class matplotlib.animation.FuncAnimation(fig, func, frames=None,init_func=None, fargs=None, save_count=None, *, cache_frame_data=True, **kwargs)

animation= FuncAnimation.FuncAnimation(
    fig=fig,
    func=update_data,
    frames=len(t),
    interval=25,
)
animation.save("animation.gif")





```
