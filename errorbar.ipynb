```python
import numpy as np
import matplotlib.pyplot as plt

models = ['Linear Regression',
          'Decision Tree',
          'Random Forest',
          'SVM',
          'KNN']

accuracy = [85, 78, 91, 87, 82]
errors = [2, 4, 1.5, 3, 2.5]

plt.figure(figsize=(10, 6))

plt.errorbar(
    models,
    accuracy,
    yerr=errors,
    fmt='o',
    capsize=6,
    color='blue',
    ecolor='red',
    markersize=8
)

plt.title('Machine Learning Model Accuracy')
plt.xlabel('ML Models')
plt.ylabel('Accuracy (%)')

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
